# 00 · Savoroute quickstart

**Goal:** explore one synthetic catalog and run the same search code used by the CLI. All results are fictional; no APIs or external models are used.

**Setup:** from the repository root, run `python -m pip install -e ".[dev,notebooks]"`. Then launch JupyterLab.

## Step 1 — Import the library

All catalog and retrieval operations use the installed `savoroute` package.

```python
from collections import Counter
from savoroute import __version__
from savoroute.data.catalog import load_catalog
from savoroute.retrieval.lexical import SearchFilters
from savoroute.services.discovery import DiscoveryService

print("Savoroute", __version__)
```

## Step 2 — Inspect canonical data and source labels

```python
catalog = load_catalog()
print("Dataset:", catalog.dataset_name, "(synthetic:", catalog.synthetic, ")")
print("Records:", len(catalog.items), dict(Counter(i.kind for i in catalog.items)))
print("Media references:", len(catalog.media))
for item in catalog.items[:3]:
    print(item.item_id, "|", item.title, "|", item.source)
assert catalog.synthetic and len(catalog.items) == 12
```

## Step 3 — Search without a model

We use deterministic BM25 scoring over the actual catalog. Scores measure token-based ranking only; they are not probabilities.

```python
service = DiscoveryService(catalog)
hits = service.search("spicy tofu noodles", filters=SearchFilters(kind="recipe"))
for rank, hit in enumerate(hits, 1):
    print(rank, hit.item.item_id, hit.item.title, "score=", round(hit.score, 3),
          "matches=", hit.matched_terms)
assert hits[0].item.item_id == "p001"
```

## Step 4 — Modify the query interactively

Try replacing `query` below. Leave `kind` set to `restaurant` or `recipe`, or use `None`.

```python
query = "chickpea salad"
kind = "recipe"
for hit in service.search(query, filters=SearchFilters(kind=kind), limit=5):
    print(hit.item.item_id, hit.item.title)
```

## Step 5 — No matches

An unmatched query returns an empty tuple. Search results are always existing catalog records.

```python
assert service.search("quasar xylophone nebula") == ()
print("No grounded matches returned, as expected.")
```
